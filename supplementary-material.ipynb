## Lowest excitations and singlet-triplet gaps from the QUEST1 dataset

```python
XC = ["LDA", "PBE", "r2SCAN", "PBE0", "r2SCAN0", "HSE12"]
XC2name = {"LDA": "LDA", "PBE": "PBE", "r2SCAN": "r$^2$SCAN", "PBE0": "PBE0", "r2SCAN0": "r$^2$SCAN0", "HSE12": "HSE12"}

for xc in XC:
    df = pd.read_csv(f'csv/QUEST1_lowest_{xc}.csv', index_col=0)
    caption = f"Lowest singlet and triplet excitations and singlet-triplet gaps for QUEST1 calculated with the UKS-OEP, SS-KS-OEP, OSS-KS-OEP, STA-KS-OEP, UKS, pa-SS-KS, pa-OSS-KS, and pa-STA-KS methods using {XC2name[xc]} XC functional as well as theoretical best estimates (TBEs)."
    display(df.style.format(precision=3).set_caption(caption).set_table_styles(TAB_STYLE))
```
